```python
from hal.hal import HardwareAwareLayout
from hal.codes.radial_codes import RadialCode

# TODO: Change this to your local path to the HAL database
PATH_TO_DATABASE = "./database/multiprocessing_test"
```

```python
# 1. Define the parameter space for the sweep
sweep_parameters = {
    "layout": ["community", "kamada_kawai"],
    "route_edge_order": ["length_asc", "length_desc", "random"],
    "node_expansion_val": [1, 2],
}

# 2. Initialize your base HardwareAwareLayout instance
# This instance provides the base Tanner graph and default settings
code = RadialCode(r=2, s=2)

# Initialize HAL with default settings
hal_base = HardwareAwareLayout(
    name="radial_code_example",
    directory_path=PATH_TO_DATABASE,
    tanner_graph=code.graph
)

# 3. Run the sweep
# This will execute 2 * 3 * 2 = 12 different configurations in parallel
results_df = hal_base.benchmark_sweep(sweep_params=sweep_parameters)

# 4. Analyze the results
print("\n--- Aggregated Sweep Results ---")
print(results_df)

# For example, find the best configuration
best_config = results_df.loc[results_df['hardware_complexity'].idxmin()]
print("\n--- Best Configuration Found ---")
print(best_config)
```
